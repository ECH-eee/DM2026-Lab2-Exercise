```python
import os
# Force single-GPU execution (avoids DataParallel VRAM overhead on Kaggle T4 x2)
os.environ["CUDA_VISIBLE_DEVICES"] = "0"
os.environ["TOKENIZERS_PARALLELISM"] = "false"

import re
import glob
import html
import json
import random
import shutil
import inspect
import numpy as np
import pandas as pd
import torch
import transformers
from packaging import version
from scipy.sparse import hstack
from scipy.optimize import minimize
from sklearn.model_selection import StratifiedKFold
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.decomposition import TruncatedSVD
from sklearn.metrics import f1_score, accuracy_score, classification_report
from datasets import Dataset
from transformers import (
    AutoTokenizer,
    AutoModelForSequenceClassification,
    TrainingArguments,
    Trainer,
    DataCollatorWithPadding,
    EarlyStoppingCallback,
)

assert version.parse(transformers.__version__) >= version.parse("4.30.0"), (
    f"Requires transformers>=4.30.0, but found {transformers.__version__}. "
    "Switch to a newer Kaggle Docker image or run: !pip install -q -U 'transformers>=4.48.0'"
)

try:
    from gensim.models import FastText
    HAS_GENSIM = True
except ImportError:
    HAS_GENSIM = False

# ==========================================================================
# KAGGLE CONFIGURATION
# ==========================================================================
HF_MODEL_ID = "j-hartmann/emotion-english-roberta-large"
OUTPUT_DIR = "/kaggle/working"
KEEP_CHECKPOINTS = False   # Keep False on Kaggle to stay well within 20 GB /kaggle/working disk limit
N_FOLDS = 5
MAX_LEN = 192
EPOCHS = 4
SEED = 42

os.makedirs(OUTPUT_DIR, exist_ok=True)

def seed_everything(seed=SEED):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    if torch.cuda.is_available():
        torch.cuda.manual_seed_all(seed)

seed_everything(SEED)

if torch.cuda.is_available():
    device = "cuda"
    gpu_name = torch.cuda.get_device_name(0)
    vram_gb = torch.cuda.get_device_properties(0).total_memory / (1024 ** 3)
    USE_BF16 = torch.cuda.is_bf16_supported()
    USE_FP16 = not USE_BF16
    print(f"Device: CUDA ({gpu_name}, {vram_gb:.1f} GB VRAM) | bf16={USE_BF16}, fp16={USE_FP16} | transformers={transformers.__version__} | gensim={HAS_GENSIM}")
else:
    device = "cpu"
    USE_BF16 = False
    USE_FP16 = False
    print(f"Warning: GPU accelerator is not enabled in Kaggle Settings! Running on CPU | transformers={transformers.__version__} | gensim={HAS_GENSIM}")
```

```python
# Auto-discover competition dataset directory under /kaggle/input
DATA_DIR = "/kaggle/input/emotion-classification"
if not os.path.exists(os.path.join(DATA_DIR, "final_posts.json")):
    matches = glob.glob("/kaggle/input/**/final_posts.json", recursive=True)
    assert len(matches) > 0, "Could not find final_posts.json under /kaggle/input. Attach the competition dataset first."
    DATA_DIR = os.path.dirname(matches[0])

print(f"Loading dataset from: {DATA_DIR}")

with open(os.path.join(DATA_DIR, "final_posts.json"), "r", encoding="utf-8") as f:
    posts_df = pd.DataFrame(json.load(f))

split_df = pd.read_csv(os.path.join(DATA_DIR, "data_identification.csv"))
emotion_df = pd.read_csv(os.path.join(DATA_DIR, "emotion.csv"))
sample_sub = pd.read_csv(os.path.join(DATA_DIR, "sampleSubmission.csv"))

df = (
    posts_df
    .merge(split_df, on="post_id", how="left")
    .merge(emotion_df, on="post_id", how="left")
)

# Dynamic assertions safe for hidden private leaderboard reruns
assert len(df) == len(split_df), f"Expected {len(split_df)} rows, got {len(df)}"
assert (df[df["set"] == "test"]["post_id"].values == sample_sub[sample_sub.columns[0]].values).all(), "Test order mismatch!"
print("Dataset loaded and verified. Shape:", df.shape)
print("Columns in final_posts.json:", posts_df.columns.tolist())
print("\nTraining label distribution:\n", df[df["set"] == "train"]["label"].value_counts())
```

```python
# ==========================================================================
# STEP 1: Dual-Branch Social-Media Pre-processing
# ==========================================================================
# - roberta_text : Natural cased text preserving Unicode emojis and [NAME]
#                  (matches j-hartmann/emotion-english-roberta-large pretraining)
# - clean_text   : Normalized text with verbalized emojis/emoticons and
#                  <USER>/<URL>/<NAME> tags for TF-IDF & FastText models
EMOTICON_MAP = {
    ":)": " [happy] ", ":-)": " [happy] ", ":D": " [joy] ", ":-D": " [joy] ",
    "😂": " [laughing joy] ", "🤣": " [laughing joy] ", "😊": " [warm happy] ",
    "😍": " [love joy] ", "❤️": " [love] ", "<3": " [love] ", "</3": " [heartbroken] ",
    "😭": " [sobbing sadness] ", "😢": " [crying sadness] ", "😞": " [sad] ",
    ":(": " [sad] ", ":-(": " [sad] ", ":'(": " [crying] ", "💔": " [heartbroken] ",
    "😡": " [furious anger] ", "🤬": " [swearing anger] ", "😠": " [angry] ",
    ">:(": " [angry] ", "😤": " [frustrated anger] ",
    "🤢": " [nauseated disgust] ", "🤮": " [vomiting disgust] ", "😒": " [disgusted] ",
    "😱": " [screaming fear] ", "😨": " [fearful] ", "😰": " [anxious fear] ",
    "😬": " [nervous fear] ",
    "😲": " [astonished surprise] ", "😳": " [shocked surprise] ", "🤯": " [mind blown surprise] ",
    "😮": " [surprised] ",
}

def verbalize_emojis(text: str) -> str:
    for k, v in EMOTICON_MAP.items():
        if k in text:
            text = text.replace(k, v)
    return text

def collapse_repetitions(text: str) -> str:
    """Collapse repetition loops and cap character/emoji elongation at 3."""
    text = re.sub(r'(\b\w+(?:\s+\w+){0,4}\b)(?:[,\s]+\1){2,}', r'\1', text, flags=re.IGNORECASE)
    text = re.sub(r'([!?.,])\1{2,}', r'\1\1', text)
    text = re.sub(r'(.)\1{3,}', r'\1\1\1', text)
    return text

def clean_roberta_text(text: str) -> str:
    """Natural social-media text tailored for RoBERTa byte-level BPE."""
    if not isinstance(text, str):
        return ""
    text = html.unescape(text).replace(r"\/", "/")
    text = re.sub(r"^\s*>+\s*", "", text)
    # Use standard subwords (@user, http) and preserve GoEmotions [NAME] without adding new special tokens
    text = re.sub(r"@<username>", "@user", text)
    text = re.sub(r"#{1,3}URL#{1,3}|\bURL\b", "http", text)
    text = re.sub(r"(@user\s*){2,}", "@user ", text)
    text = re.sub(r"(http\s*){2,}", "http ", text)
    text = collapse_repetitions(text)
    text = re.sub(r"\s+", " ", text).strip()
    return text

def clean_lexical_text(text: str) -> str:
    """Normalized text with verbalized emojis/emoticons for TF-IDF & FastText."""
    if not isinstance(text, str):
        return ""
    text = html.unescape(text).replace(r"\/", "/")
    text = re.sub(r"^\s*>+\s*", "", text)
    text = re.sub(r"@<username>", "<USER>", text)
    text = re.sub(r"#{1,3}URL#{1,3}|\bURL\b", "<URL>", text)
    text = re.sub(r"\[NAME\]", "<NAME>", text)
    text = re.sub(r"(<USER>\s*){2,}", "<USER> ", text)
    text = re.sub(r"(<URL>\s*){2,}", "<URL> ", text)
    # Collapse repeated emojis/chars BEFORE expanding into multi-word tags
    text = collapse_repetitions(text)
    text = verbalize_emojis(text)
    text = re.sub(r"\s+", " ", text).strip()
    return text

# Detect any extra metadata columns in final_posts.json besides post_id and text
meta_cols = [c for c in posts_df.columns if c not in {"post_id", "text"}]

def build_roberta_input(row) -> str:
    text = row["roberta_base_text"]
    if meta_cols:
        meta_parts = [f"{c}={row[c]}" for c in meta_cols if pd.notnull(row[c]) and str(row[c]).strip() != ""]
        if meta_parts:
            return "[Meta: " + ", ".join(meta_parts) + "] " + text
    return text

df["clean_text"] = df["text"].apply(clean_lexical_text)
df["roberta_base_text"] = df["text"].apply(clean_roberta_text)
df["roberta_text"] = df.apply(build_roberta_input, axis=1)

# Separate train and test
train_df = df[df["set"] == "train"].copy()
test_df = df[df["set"] == "test"].copy()

# Remove placeholder-only rows with zero semantic words from train
has_word_mask = train_df["clean_text"].str.replace(r"<USER>|<URL>|<NAME>|\W+", "", regex=True).str.len() > 0
print(f"Dropping {(~has_word_mask).sum()} empty/placeholder-only rows from train.")
train_df = train_df[has_word_mask].copy()

# 1. Deduplicate exact (clean_text, label) pairs
before_dedup = len(train_df)
train_df = train_df.drop_duplicates(subset=["clean_text", "label"]).reset_index(drop=True)
print(f"Deduplicated exact (text, label) rows: {before_dedup} -> {len(train_df)}")

# 2. Drop ambiguous posts that appear with conflicting labels in train
label_nunique = train_df.groupby("clean_text")["label"].transform("nunique")
conflict_cnt = int((label_nunique > 1).sum())
if conflict_cnt > 0:
    train_df = train_df[label_nunique == 1].reset_index(drop=True)
    print(f"Dropped {conflict_cnt} rows with contradictory labels -> {len(train_df)} clean train rows.")
```

```python
LABELS = sorted(train_df["label"].unique().tolist())
label2id = {lbl: idx for idx, lbl in enumerate(LABELS)}
id2label = {idx: lbl for idx, lbl in enumerate(LABELS)}

train_df["label_id"] = train_df["label"].map(label2id)
print("Label mapping:", label2id)

class_counts = np.bincount(train_df["label_id"].values)
class_priors = class_counts / class_counts.sum()
print("Class priors:", dict(zip(LABELS, np.round(class_priors, 4))))

skf = StratifiedKFold(n_splits=N_FOLDS, shuffle=True, random_state=SEED)
train_df["fold"] = -1
for fold, (_, val_idx) in enumerate(skf.split(train_df, train_df["label_id"])):
    train_df.loc[val_idx, "fold"] = fold
```

```python
# ==========================================================================
# STEP 2: Word+Char TF-IDF & Unsupervised IDF-Weighted FastText Embeddings
# ==========================================================================
tfidf_oof_probs = np.zeros((len(train_df), len(LABELS)))
tfidf_test_probs = np.zeros((len(test_df), len(LABELS)))

emb_oof_probs = np.zeros((len(train_df), len(LABELS)))
emb_test_probs = np.zeros((len(test_df), len(LABELS)))

# 12 dense similarity features: 6 document-to-centroid + 6 peak-word cosine similarities
sim_oof_feats = np.zeros((len(train_df), len(LABELS) * 2), dtype=np.float32)
sim_test_feats = np.zeros((len(test_df), len(LABELS) * 2), dtype=np.float32)

if HAS_GENSIM:
    all_tokenized = [t.lower().split() for t in pd.concat([train_df["clean_text"], test_df["clean_text"]])]
    ft_model = FastText(
        sentences=all_tokenized,
        vector_size=128,
        window=5,
        min_count=2,
        min_n=3,
        max_n=6,
        sg=1,
        epochs=30,
        seed=SEED,
    )
    ft_wv = ft_model.wv

    def get_idf_doc_and_tok_vecs(texts, idf_lookup, default_idf):
        dim = ft_wv.vector_size
        doc_vecs = np.zeros((len(texts), dim), dtype=np.float32)
        tok_vecs_list = []
        for i, text in enumerate(texts):
            tokens = text.lower().split()
            if not tokens:
                tok_vecs_list.append(np.zeros((1, dim), dtype=np.float32))
                continue
            weights = np.array([idf_lookup.get(tok, default_idf) for tok in tokens], dtype=np.float32)
            vecs = np.vstack([ft_wv[tok] for tok in tokens])
            norms = np.linalg.norm(vecs, axis=1, keepdims=True) + 1e-9
            tok_vecs_list.append(vecs / norms)
            weighted = (vecs * weights[:, None]).sum(axis=0) / (weights.sum() + 1e-9)
            doc_vecs[i] = weighted / (np.linalg.norm(weighted) + 1e-9)
        return doc_vecs, tok_vecs_list
else:
    base_svd_vec = TfidfVectorizer(ngram_range=(1, 2), sublinear_tf=True, min_df=2, max_features=20000)
    all_sparse = base_svd_vec.fit_transform(pd.concat([train_df["clean_text"], test_df["clean_text"]]))
    svd_model = TruncatedSVD(n_components=128, random_state=SEED)
    all_dense = svd_model.fit_transform(all_sparse).astype(np.float32)
    all_dense = all_dense / (np.linalg.norm(all_dense, axis=1, keepdims=True) + 1e-9)
    trn_svd_vecs = all_dense[:len(train_df)]
    tst_svd_vecs = all_dense[len(train_df):]

for fold in range(N_FOLDS):
    trn_mask = (train_df["fold"] != fold).values
    val_mask = (train_df["fold"] == fold).values
    y_trn = train_df.loc[trn_mask, "label_id"].values

    # Model 1 — Word (1, 2) + Character (3, 5) TF-IDF
    word_vec = TfidfVectorizer(
        ngram_range=(1, 2),
        min_df=2,
        sublinear_tf=True,
        max_features=20000,
        token_pattern=r"(?u)\b\w+\b|<USER>|<URL>|<NAME>",
    )
    char_vec = TfidfVectorizer(
        analyzer="char_wb",
        ngram_range=(3, 5),
        min_df=3,
        sublinear_tf=True,
        max_features=25000,
    )
    X_trn_tfidf = hstack([word_vec.fit_transform(train_df.loc[trn_mask, "clean_text"]),
                          char_vec.fit_transform(train_df.loc[trn_mask, "clean_text"])])
    X_val_tfidf = hstack([word_vec.transform(train_df.loc[val_mask, "clean_text"]),
                          char_vec.transform(train_df.loc[val_mask, "clean_text"])])
    X_tst_tfidf = hstack([word_vec.transform(test_df["clean_text"]),
                          char_vec.transform(test_df["clean_text"])])

    clf_tfidf = LogisticRegression(C=2.0, class_weight="balanced", max_iter=1000, random_state=SEED)
    clf_tfidf.fit(X_trn_tfidf, y_trn)
    tfidf_oof_probs[val_mask] = clf_tfidf.predict_proba(X_val_tfidf)
    tfidf_test_probs += clf_tfidf.predict_proba(X_tst_tfidf) / N_FOLDS

    # Model 2 — Unsupervised FastText + IDF Document Embeddings & Emotion Centroid Similarities
    if HAS_GENSIM:
        idf_vec = TfidfVectorizer(lowercase=True, token_pattern=r"(?u)\S+")
        idf_vec.fit(train_df.loc[trn_mask, "clean_text"])
        idf_lookup = dict(zip(idf_vec.get_feature_names_out(), idf_vec.idf_))
        default_idf = float(np.median(idf_vec.idf_))

        X_trn_emb, _ = get_idf_doc_and_tok_vecs(train_df.loc[trn_mask, "clean_text"], idf_lookup, default_idf)
        X_val_emb, val_tok_vecs = get_idf_doc_and_tok_vecs(train_df.loc[val_mask, "clean_text"], idf_lookup, default_idf)
        X_tst_emb, tst_tok_vecs = get_idf_doc_and_tok_vecs(test_df["clean_text"], idf_lookup, default_idf)
    else:
        X_trn_emb = trn_svd_vecs[trn_mask]
        X_val_emb = trn_svd_vecs[val_mask]
        X_tst_emb = tst_svd_vecs

    clf_emb = LogisticRegression(C=2.0, class_weight="balanced", max_iter=1000, random_state=SEED)
    clf_emb.fit(X_trn_emb, y_trn)
    emb_oof_probs[val_mask] = clf_emb.predict_proba(X_val_emb)
    emb_test_probs += clf_emb.predict_proba(X_tst_emb) / N_FOLDS

    # Emotion class centroids strictly on training fold
    centroids = np.vstack([X_trn_emb[y_trn == c].mean(axis=0) for c in range(len(LABELS))])
    centroids = centroids / (np.linalg.norm(centroids, axis=1, keepdims=True) + 1e-9)

    sim_oof_feats[val_mask, :len(LABELS)] = X_val_emb @ centroids.T
    sim_test_feats[:, :len(LABELS)] += (X_tst_emb @ centroids.T) / N_FOLDS

    if HAS_GENSIM:
        sim_oof_feats[val_mask, len(LABELS):] = np.vstack([(tv @ centroids.T).max(axis=0) for tv in val_tok_vecs])
        sim_test_feats[:, len(LABELS):] += np.vstack([(tv @ centroids.T).max(axis=0) for tv in tst_tok_vecs]) / N_FOLDS
    else:
        sim_oof_feats[val_mask, len(LABELS):] = sim_oof_feats[val_mask, :len(LABELS)]
        sim_test_feats[:, len(LABELS):] = sim_test_feats[:, :len(LABELS)]

y_true = train_df["label_id"].values
tfidf_macro_f1 = f1_score(y_true, np.argmax(tfidf_oof_probs, axis=-1), average="macro")
emb_macro_f1 = f1_score(y_true, np.argmax(emb_oof_probs, axis=-1), average="macro")
print(f"[CPU Model 1] {N_FOLDS}-Fold Word+Char TF-IDF OOF Macro F1 : {tfidf_macro_f1:.4f}")
print(f"[CPU Model 2] {N_FOLDS}-Fold IDF-FastText/SVD OOF Macro F1 : {emb_macro_f1:.4f}")
```

```python
# ==========================================================================
# STEP 3: Kaggle RoBERTa-large Setup (j-hartmann/emotion-english-roberta-large)
# ==========================================================================
# Check if j-hartmann/emotion-english-roberta-large is attached offline under /kaggle/input;
# otherwise fall back to Hugging Face Hub (works when Kaggle Internet is turned ON).
cfg_matches = []
for root, _, files in os.walk("/kaggle/input", followlinks=True):
    if "config.json" in files:
        cfg_matches.append(os.path.join(root, "config.json"))

roberta_matches = [
    p for p in cfg_matches
    if "roberta" in p.lower() or "emotion-english" in p.lower()
]
if roberta_matches:
    MODEL_PATH = os.path.dirname(roberta_matches[0])
    IS_LOCAL_MODEL = True
    print(f"Found attached offline RoBERTa-large model under /kaggle/input: {MODEL_PATH}")
elif cfg_matches:
    MODEL_PATH = os.path.dirname(cfg_matches[0])
    IS_LOCAL_MODEL = True
    print(f"Found attached offline model under /kaggle/input: {MODEL_PATH}")
else:
    MODEL_PATH = HF_MODEL_ID
    IS_LOCAL_MODEL = False
    print(f"No local model under /kaggle/input; loading '{MODEL_PATH}' from Hugging Face Hub (requires Kaggle Internet ON).")

try:
    tokenizer = AutoTokenizer.from_pretrained(
        MODEL_PATH,
        local_files_only=IS_LOCAL_MODEL,
    )
except OSError as exc:
    raise RuntimeError(
        f"Failed to load '{MODEL_PATH}'. If running with Kaggle Internet OFF, attach "
        f"'{HF_MODEL_ID}' as a Kaggle Model/Dataset input first.\nOriginal error: {exc}"
    ) from exc

def tokenize_fn(batch):
    return tokenizer(
        batch["roberta_text"],
        truncation=True,
        max_length=MAX_LEN,
        return_token_type_ids=False,
    )

def compute_metrics(eval_pred):
    logits, labels = eval_pred
    preds = np.argmax(logits, axis=-1)
    return {
        "macro_f1": f1_score(labels, preds, average="macro"),
        "weighted_f1": f1_score(labels, preds, average="weighted"),
        "accuracy": accuracy_score(labels, preds),
    }
```

```python
# ==========================================================================
# STEP 3 (Continued): 5-Fold RoBERTa-large Training
# ==========================================================================
def build_roberta_6class_model(model_path, labels, id2label, label2id, is_local):
    """Load Hartmann's 7-class emotion checkpoint and slice matching emotion head weights into 6 classes."""
    model = AutoModelForSequenceClassification.from_pretrained(
        model_path,
        local_files_only=is_local,
    )
    model = model.float()

    src_label2id = {k.lower(): int(v) for k, v in model.config.label2id.items()}
    old_out = model.classifier.out_proj
    new_out = torch.nn.Linear(old_out.in_features, len(labels), bias=(old_out.bias is not None))
    new_out = new_out.to(dtype=old_out.weight.dtype, device=old_out.weight.device)

    matched = []
    with torch.no_grad():
        for new_idx, lbl in enumerate(labels):
            old_idx = src_label2id.get(lbl.lower())
            if old_idx is not None and old_idx < old_out.out_features:
                new_out.weight[new_idx].copy_(old_out.weight[old_idx])
                if old_out.bias is not None:
                    new_out.bias[new_idx].copy_(old_out.bias[old_idx])
                matched.append(f"{lbl}({old_idx}->{new_idx})")

    model.classifier.out_proj = new_out
    model.num_labels = len(labels)
    model.config.num_labels = len(labels)
    model.config.id2label = id2label
    model.config.label2id = label2id
    model.config.problem_type = "single_label_classification"
    return model, matched

roberta_oof_probs = np.zeros((len(train_df), len(LABELS)))
roberta_test_probs = np.zeros((len(test_df), len(LABELS)))

test_ds = Dataset.from_pandas(test_df[["roberta_text"]]).map(tokenize_fn, batched=True)
data_collator = DataCollatorWithPadding(tokenizer=tokenizer)

IS_LARGE = "large" in MODEL_PATH.lower()
BATCH_SIZE = 4 if IS_LARGE else 8
GRAD_ACCUM = 4 if IS_LARGE else 2  # Effective batch size = 16
LR = 1.5e-5 if IS_LARGE else 2e-5

# Ensure compatibility across transformers v4.48+ and v5.x
ta_params = inspect.signature(TrainingArguments.__init__).parameters
eval_strat_key = "eval_strategy" if "eval_strategy" in ta_params else "evaluation_strategy"
tr_params = inspect.signature(Trainer.__init__).parameters
trainer_tok_kwargs = {"processing_class": tokenizer} if "processing_class" in tr_params else {"tokenizer": tokenizer}

for fold in range(N_FOLDS):
    print(f"\n========== Training RoBERTa-large Fold {fold + 1} / {N_FOLDS} ==========")
    trn_split = train_df[train_df["fold"] != fold].reset_index(drop=True)
    val_split = train_df[train_df["fold"] == fold].reset_index(drop=True)
    val_indices = train_df[train_df["fold"] == fold].index.values

    trn_ds = Dataset.from_pandas(trn_split[["roberta_text", "label_id"]].rename(columns={"label_id": "labels"})).map(tokenize_fn, batched=True)
    val_ds = Dataset.from_pandas(val_split[["roberta_text", "label_id"]].rename(columns={"label_id": "labels"})).map(tokenize_fn, batched=True)

    model, matched_heads = build_roberta_6class_model(
        MODEL_PATH, LABELS, id2label, label2id, IS_LOCAL_MODEL
    )
    if fold == 0:
        print(f"Preserved pretrained classifier weights for {len(matched_heads)}/{len(LABELS)} classes: {matched_heads}")

    ckpt_dir = os.path.join(OUTPUT_DIR, f"checkpoints_fold_{fold}")
    total_steps = max(1, (len(trn_ds) // (BATCH_SIZE * GRAD_ACCUM)) * EPOCHS)
    warmup_steps = max(1, int(0.1 * total_steps))

    ta_kwargs = {
        "output_dir": ckpt_dir,
        eval_strat_key: "epoch",
        "save_strategy": "epoch",
        "learning_rate": LR,
        "lr_scheduler_type": "cosine",
        "per_device_train_batch_size": BATCH_SIZE,
        "gradient_accumulation_steps": GRAD_ACCUM,
        "per_device_eval_batch_size": 16,
        "num_train_epochs": EPOCHS,
        "weight_decay": 0.01,
        "warmup_steps": warmup_steps,
        "bf16": USE_BF16,
        "fp16": USE_FP16,
        "gradient_checkpointing": IS_LARGE,
        "load_best_model_at_end": True,
        "metric_for_best_model": "macro_f1",
        "greater_is_better": True,
        "save_total_limit": 1,
        "dataloader_num_workers": 2,
        "logging_steps": 50,
        "report_to": "none",
    }
    if IS_LARGE and "gradient_checkpointing_kwargs" in ta_params:
        ta_kwargs["gradient_checkpointing_kwargs"] = {"use_reentrant": False}

    training_args = TrainingArguments(**ta_kwargs)

    trainer = Trainer(
        model=model,
        args=training_args,
        train_dataset=trn_ds,
        eval_dataset=val_ds,
        data_collator=data_collator,
        compute_metrics=compute_metrics,
        callbacks=[EarlyStoppingCallback(early_stopping_patience=2)],
        **trainer_tok_kwargs,
    )

    trainer.train()

    val_logits = trainer.predict(val_ds).predictions
    roberta_oof_probs[val_indices] = torch.softmax(torch.tensor(val_logits), dim=-1).numpy()

    fold_test_logits = trainer.predict(test_ds).predictions
    roberta_test_probs += torch.softmax(torch.tensor(fold_test_logits), dim=-1).numpy() / N_FOLDS

    if KEEP_CHECKPOINTS:
        best_dir = os.path.join(OUTPUT_DIR, f"roberta_best_fold_{fold}")
        trainer.save_model(best_dir)
        tokenizer.save_pretrained(best_dir)

    del model, trainer
    if torch.cuda.is_available():
        torch.cuda.empty_cache()
    shutil.rmtree(ckpt_dir, ignore_errors=True)
```

```python
# ==========================================================================
# STEP 4: Soft-Voting Ensemble & Per-Class Probability Multiplier Tuning
# ==========================================================================
y_true = train_df["label_id"].values
folds = train_df["fold"].values

roberta_f1 = f1_score(y_true, np.argmax(roberta_oof_probs, axis=-1), average="macro")
print(f"Model 1 (Word+Char TF-IDF)              OOF Macro F1 : {tfidf_macro_f1:.4f}")
print(f"Model 2 (IDF-FastText/SVD)              OOF Macro F1 : {emb_macro_f1:.4f}")
print(f"Model 3 (RoBERTa-large-emotions  )      OOF Macro F1 : {roberta_f1:.4f}")

# Weighted Soft-Voting Blend across the 3 models
blend_oof_probs = 0.85 * roberta_oof_probs + 0.10 * tfidf_oof_probs + 0.05 * emb_oof_probs
blend_test_probs = 0.85 * roberta_test_probs + 0.10 * tfidf_test_probs + 0.05 * emb_test_probs
blend_f1 = f1_score(y_true, np.argmax(blend_oof_probs, axis=-1), average="macro")
print(f"Soft-Voting Blend (0.85/0.10/0.05) OOF Macro F1: {blend_f1:.4f}")

if blend_f1 >= roberta_f1:
    chosen_oof_probs, chosen_test_probs, base_f1 = blend_oof_probs, blend_test_probs, blend_f1
    print("Selected: 3-Model Soft-Voting Blend")
else:
    chosen_oof_probs, chosen_test_probs, base_f1 = roberta_oof_probs, roberta_test_probs, roberta_f1
    print("Selected: RoBERTa-large-emotions   Alone")

# Out-of-Fold Per-Class Probability Multiplier Tuning for Macro F1
def optimize_class_multipliers(oof_p, y_arr, fold_arr):
    n_classes = oof_p.shape[1]
    fold_weights = []
    nested_preds = np.zeros(len(y_arr), dtype=int)
    for val_fold in np.unique(fold_arr):
        trn_m = fold_arr != val_fold
        val_m = fold_arr == val_fold
        def loss_fn(weights):
            w = np.clip(weights, 0.25, 4.0)
            return -f1_score(y_arr[trn_m], np.argmax(oof_p[trn_m] * w, axis=-1), average="macro")
        res = minimize(loss_fn, x0=np.ones(n_classes), method="Nelder-Mead", options={"maxiter": 1000})
        w_opt = np.clip(res.x, 0.25, 4.0)
        fold_weights.append(w_opt)
        nested_preds[val_m] = np.argmax(oof_p[val_m] * w_opt, axis=-1)
    nested_f1 = f1_score(y_arr, nested_preds, average="macro")
    return np.mean(fold_weights, axis=0), nested_f1, nested_preds

best_multipliers, tuned_nested_f1, tuned_nested_preds = optimize_class_multipliers(chosen_oof_probs, y_true, folds)
if tuned_nested_f1 > base_f1:
    print(f"Applied OOF Class Multipliers: {base_f1:.4f} -> {tuned_nested_f1:.4f}")
    print("Learned Multipliers:", dict(zip(LABELS, np.round(best_multipliers, 3))))
    final_oof_preds = tuned_nested_preds
    final_test_preds = np.argmax(chosen_test_probs * best_multipliers, axis=-1)
else:
    print(f"Kept unscaled ensemble probabilities (OOF {tuned_nested_f1:.4f} <= {base_f1:.4f})")
    final_oof_preds = np.argmax(chosen_oof_probs, axis=-1)
    final_test_preds = np.argmax(chosen_test_probs, axis=-1)

np.save(os.path.join(OUTPUT_DIR, "roberta_oof_probs.npy"), roberta_oof_probs)
np.save(os.path.join(OUTPUT_DIR, "roberta_test_probs.npy"), roberta_test_probs)
np.save(os.path.join(OUTPUT_DIR, "tfidf_oof_probs.npy"), tfidf_oof_probs)
np.save(os.path.join(OUTPUT_DIR, "tfidf_test_probs.npy"), tfidf_test_probs)
np.save(os.path.join(OUTPUT_DIR, "emb_oof_probs.npy"), emb_oof_probs)
np.save(os.path.join(OUTPUT_DIR, "emb_test_probs.npy"), emb_test_probs)

print(f"\nFinal OOF Macro F1 : {f1_score(y_true, final_oof_preds, average='macro'):.4f}")
print(f"Final OOF Accuracy : {accuracy_score(y_true, final_oof_preds):.4f}\n")
print(classification_report(y_true, final_oof_preds, target_names=LABELS, digits=4))
```
